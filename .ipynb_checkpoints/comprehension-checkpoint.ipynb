```python
vendas = [100, 250, 80, 500, 120]
```

```python
# Uma lista com os valores dobrados.

lista = [
    x**2
    for x in vendas
]
print(lista)
```

```python
# Um set com os valores únicos.

conjunto = {c for c in vendas}
print(conjunto)
```

```python
# Um dicionário onde a chave seja o valor e o valor seja o triplo dele.
dicio = {
    d:d*3
    for d in vendas
}
print(dicio)
```

```python
# Um generator que produza os quadrados.
generator = (x for x in vendas)
print(f'{generator}\n')
type(generator)
```
