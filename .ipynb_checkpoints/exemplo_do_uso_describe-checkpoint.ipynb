```python
import pandas as pd

# 1️⃣ Carregar dados
df = pd.read_csv('vendas.csv', parse_dates=['mes'])

# 2️⃣ Gerar estatísticas resumidas
resumo = df['valor_venda'].describe()

print("Resumo das vendas mensais:")
print(resumo)

# 3️⃣ Usar estatísticas para decisões
#   a) Identificar se há vendas muito abaixo da média (outliers)
outlier_threshold = resumo['mean'] - 2 * resumo['std']
baixo = df[df['valor_venda'] < outlier_threshold]
print("\nVendas abaixo de 2 desvios padrão:")
print(baixo)

#   b) Planejar estoque baseado no 75º percentil
estoque_minimo = resumo['75%']
print(f"\nRecomenda-se estoque mínimo de {estoque_minimo:.2f} unidades por mês.")
```

O que acontece?
1. Carregamento: read_csv lê o arquivo e converte a coluna mes para datetime.
2. describe(): Calcula count, mean, std, min, 25%, 50%, 75%, max.
3. Outlier Detection: Usamos média e desvio padrão para flagar vendas muito baixas.
4. Planejamento de Estoque: O 75º percentil indica que 75 % das vendas estão abaixo desse valor; usar isso como base para um estoque mínimo evita rupturas.
