## Exercício 6: Conta bancária com proteção contra depósitos e sobregiro

Ele introduz o conceito de gerenciamento de estado dentro de uma classe, um padrão encontrado em diversos lugares, desde componentes de interface do usuário e dispositivos IoT até objetos de jogos e mecanismos de fluxo de trabalho.

Estado atual: DESLIGADO
