#### Exercício 2

Crie duas variáveis, numero1 e numero2, e atribua a elas valores inteiros. Calcule a soma, subtração, multiplicação e divisão dessas variáveis e imprima os resultados.

```python
# Solução
Num1 = 2
Num2 = 3

# soma
def soma():
    return(Num1 + Num2)
# subtração
def subtracao():
    return(Num1 - Num2)
# multiplicação
def multiplicacao():
    return(Num1 * Num2)
# divisão
def divisao():
    return(Num1 / Num2)

print(f'soma: {soma()}')
print(f'subtracao: {subtracao()}')
print(f'multiplicacao: {multiplicacao()}')
print(f'divisao: {divisao():.2f}')
```

#### Exercício 3 

Qual é a principal diferença entre uma variável de escopo local e uma de escopo global?

```python
# Solução
# Uma variável de escopo local só funcionará dentro de uma função, logo para usa-la deve-se requisitar a função
# Já uma função global quando pode ser requisitada em qualquer parte do código, desde que tenha sido definida.
#exemplo:
nome = "Renan"

def saudacao():
    nome_saudacao = "Uchoa"
    print(f'futuro eng de dados: {nome_saudacao}')

print(f'\nFora de função: \n{nome}')
print(f'\nDentro da função: ')
saudacao()
```

#### Exercício 4

Crie uma variável chamada saldo com o valor 500.50 (float). Em seguida, crie uma variável saque com o valor 200.25 (float). Subtraia o saque do saldo e imprima o saldo final formatado para duas casas decimais.

#### Exercício 8

#### Exercício 29

Crie dois conjuntos: 

- set_a = {1, 2, 3, 4}
- set_b = {3, 4, 5, 6}

Encontre e imprima a interseção entre os dois conjuntos (os elementos que estão em ambos).

```python
# Solução
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(f'A interseção: {set_a.intersection(set_b)}')
print(f'A união {set_a.union(set_b)}')
print(f'Outros: {set_a.difference(set_b)}')
```

#### Exercício 30

Escreva um programa que peça ao usuário para digitar sua altura em metros (ex: 1.75) e seu peso em quilogramas (ex: 68.5). Calcule o Índice de Massa Corporal (IMC) usando a fórmula IMC = peso / (altura * altura) e imprima o resultado formatado com duas casas decimais.

```python
# Solução
altura = float(input('Qual sua altura: '))
peso = float(input("Qual seu peso: "))

IMC = peso / (altura * altura)

if IMC < 18.5:
    print('Baixo peso! Procure um nutri!')
elif IMC < 24.99:
    print('Peso saudável')
elif IMC < 30:
    print('Sobrepeso, Procure um nutri!')
elif IMC < 35:
    print('Obesidade grau 1, Procure um nutri!')
elif IMC < 40:
    print('Obesidade grau 2, Procure um nutri!')
else:
    print('Obesidade grau 3, Procure um nutri!')
    
```
