#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
# Solução

# função que classifica os triangulos
def classificacao_triangulo (a,b,c):
    if (a == b) and (b == c) and (c == a):
        print('Todos os lados são iguais, logo Equilátero')
    elif (a == b) or (b == c) or (c == a):
        print('Tem 2 lados iguais, logo Isóceles')
    else :
        print('Todos os lados são diferentes, logo Escaleno')

# lista que recebe o valor de cada lado
lados = []

# coletor de lados
for n in range(3):
    lado = int(input(f'Digite o valor do lado {n+1}: '))
    lados.append(lado)
print('- '*10)


# Resultados
n=1
for l in lados:
    print(f'lado {n}: {l} ')
    n += 1
print('- '*10)

#classificacao_triangulo(lados[0],lados[1],lados[2])
classificacao_triangulo(*lados) #uma outra forma de fazer o mesmo resultado.
```

```python
# Solução
def dsa_calcula_imc (peso,altura):
    IMC = peso / (altura * altura)

    if IMC < 18.5:
        print('Baixo peso! Procure um nutri!')
    elif IMC < 24.99:
        print('Peso saudável')
    elif IMC < 30:
        print('Sobrepeso, Procure um nutri!')
    elif IMC < 35:
        print('Obesidade grau 1, Procure um nutri!')
    elif IMC < 40:
        print('Obesidade grau 2, Procure um nutri!')
    else:
        print('Obesidade grau 3, Procure um nutri!')


peso = float(input("Qual seu peso(ex: 80): "))
altura = float(input('Qual sua altura(ex: 1.78): '))
dsa_calcula_imc (peso,altura)
```

#### Exercício 7

Crie uma função que receba uma lista de números inteiros e retorne um dicionário contendo a contagem de quantos números são pares e quantos são ímpares.

```python
# Solução
numeros = list(range(1,20))

def par_impar(lista_numeros):
    pares = list(filter(lambda x: x%2==0,lista_numeros))
    impares = list(filter(lambda x: x%2!=0, lista_numeros))

    print(f'\nnúmeros pares:{pares}')
    print(f'\nnúmeros impares:{impares}')
    contagem = {'pares':len(pares),'impares':len(impares)}
    print()
    #print(contagem)
    return contagem

par_impar(numeros)
```

```python
# Solução 02
numeros = list(range(1,20))

def par_impar(lista_numeros):
    contagem = {'pares': 0 ,'impares': 0}

    for i in lista_numeros:
        if i %2 == 0:
            contagem['pares'] += 1
        else:
            contagem['impares'] += 1
            
    return contagem

par_impar(numeros)
```

#### Exercício 10

Crie um jogo onde o computador escolhe um número secreto entre 1 e 20. O jogador tem 5 tentativas para adivinhar. A cada tentativa, o programa informa se o palpite foi muito alto ou muito baixo. Se o jogador acertar, o loop deve ser interrompido com uma mensagem de vitória.

```python
# Solução
import random

machine = random.randint(1,20)
#print(f'numero aleatorio: {machine}')
tentativas = 5

while tentativas > 0:
    try:
        jogador = int(input('Qual o número para ser adivinhado?'))
        if jogador == machine:
            print('Você acertou!! parabens!!')
            print(f'o numero da máquina foi: {machine}')
            break
        else:
            tentativas -= 1
            print(f'errou! Você tem {tentativas} tentativas')
            if machine > jogador:
                print('tente um número maior')
            else:
                print('tente um número menor')
            print('- '*10)
    except:
        print('Somente números válidos por favor!')
        continue
if tentativas == 0:
    print(f'o numero da máquina foi: {machine}')
```
