# Pseudocódigo para o jogo
##  O objetivo é derrotar o oponente escolhendo um de três objetos, cada qual com uma vantagem sobre um objeto e uma desvantagem sobre o outro.
### As regras de vitória são as seguintes:
- Pedra vence a Tesoura (quebrando-a).
- Tesoura vence o Papel (cortando-o).
- Papel vence a Pedra (cobrindo-a)


## Início

#### Escolha um número:
1. pedra
2. papel
3. tesoura

#### lógica:
- maquina = um número aleatório de 1 a 3
- input para escolher um número
- sequencia de condicionas seguindo as regras:
    * if maquina == 1 and jogador == 3:
        - print("você perdeu!")
    * elif maquina == 2 and jogador == 1:
        - print("você perdeu!")
    * elif maquina == 3 and jogador == 2:
        - print("você perdeu!")
    * elif maquina == jogador:
        - print('Deu empate!')
    * else:
        - print("você ganhou!")
