- Vamos jogar pedra, papel, tesoura, lagarto, spock.
- Criado pego personagem Sheldon Cooper

## Regras do jogo:
- Cada jogador deve escolher uma das opções:
- pedra, papel, tesoura, lagarto ou spock

## Ordem de poder:
##### Nas palavras do criador do jogo:
"A tesoura corta o papel, o papel cobre a pedra, a pedra esmaga o lagarto, o lagarto envenena o Spock, o Spock quebra a tesoura, a tesoura decapita o lagarto, o lagarto come o papel, o papel refuta o Spock, o Spock vaporiza a pedra e, como sempre, a pedra amassa a tesoura."

```python
# --- opções válidas ---
opcoes_validas = ('pedra','papel','tesoura','lagarto','spock')
print(f'Opções válidas: {opcoes_validas}')
print('- ' * 33)

# --- Coleta de dados ---

jogador_1_input = input('Jogador 1, digite sua jogada: ') 
jogador_2_input = input('Jogador 2, digite sua jogada: ')
print('- ' * 33)

# --- Tratamento de dados ---
jogada_1 = jogador_1_input.lower().strip()
jogada_2 = jogador_2_input.lower().strip()

print(f'O jogador 1 escolher: {jogada_1}')
print(f'O jogador 2 escolher: {jogada_2}')
print('- ' * 33)

# --- Lógica do jogo ---

if jogada_1 not in opcoes_validas or jogada_2 not in opcoes_validas:
#if opcoes_validas.count(jogada_jogador1) == 0 or opcoes_validas.count(jogada_jogador2) == 0:
#if {jogada_1, jogada_2} - set(opcoes_validas):
#if not all(jogada in opcoes_validas for jogada in [jogada_1, jogada_2]):
#if len(list(filter(lambda x: x not in opcoes_validas, [jogada_1, jogada_2]))) > 0:
    print('Somente jogadas válidas, por favor! Um dos jogadores erro, tente novamente.')
    print(f'Opções válidas: {opcoes_validas}')
    print('- ' * 33)

# --- Empate ---
elif jogada_1 == jogada_2:
    print('Jogadas Iguais, deu empate!')


# --- Vitória jogador 1 ---
elif (jogada_1 == 'pedra' and jogada_2 == 'tesoura') or \
    (jogada_1 == 'papel' and jogada_2 == 'pedra') or \
    (jogada_1 == 'pedra' and jogada_2 == 'lagarto') or \
    (jogada_1 == 'lagarto' and jogada_2 == 'spock') or \
    (jogada_1 == 'tesoura' and jogada_2 == 'lagarto') or \
    (jogada_1 == 'lagarto' and jogada_2 == 'papel') or \
    (jogada_1 == 'papel' and jogada_2 == 'spock') or \
    (jogada_1 == 'spock' and jogada_2 == 'pedra'):
    print('Jogador 1 ganhou')

# --- Vditória jogador 2 ---
else:
    print('Jogador 2 ganhou!')
```
